The equation is: $$y^2=x^3+15347898055371580590890576721314318823207531963035637503096292x+7444386449934505970367865204569124728350661870959593404279615$$ with modulo $$p=17676318486848893030961583018778670610489016512983351739677143$$
$$P=(1619092589586542907492569170434842128165755668543894279235270, 3436949547626524920645513316569700140535482973634182925459687)$$

```python
from datetime import datetime
from smart_attack import smart_attack
```

```python
p = 17676318486848893030961583018778670610489016512983351739677143; p
```

```python
p.nbits()
```

```python
E = EllipticCurve(GF(p), [15347898055371580590890576721314318823207531963035637503096292, 7444386449934505970367865204569124728350661870959593404279615]); E
```

```python
E.cardinality() == p
```

```python
P = E(1619092589586542907492569170434842128165755668543894279235270, 3436949547626524920645513316569700140535482973634182925459687); P
```

```python
n = randint(1, p-1)
```

```python
Q = n*P; Q
```

```python
time_start = datetime.now()

m = smart_attack(E, P, Q)

time_end = datetime.now()
time_taken = time_end - time_start
print("The secret is: " + str(m))
print("Time taken for smart attack to solve 204-bit ECDLP: " + str(time_taken))
```
