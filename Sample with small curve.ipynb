This is an example shown on the original smart attack in 1999, with the curve $$y^2=x^3-4x^2-128x-432$$ with modulo $$p=43$$

```python
from datetime import datetime
from smart_attack import smart_attack
```

```python
p = 43; p
```

```python
E = EllipticCurve(GF(p), [0, -4, 0, -128, -432]); E
```

```python
E.cardinality() == p
```

```python
P = E(0, 16); P
```

```python
Q = E(12, 1); Q
```

```python
Eq = EllipticCurve(pAdicField(p, 8), [0, -4, 0, -128, -432]); Eq
```

```python
time_start = datetime.now()

m = smart_attack(E, P, Q)

time_end = datetime.now()
time_taken = time_end - time_start
print("The secret is: " + str(m))
print("Time taken for smart attack to solve 43-element curve (paper example): " + str(time_taken))
```

```python
Q == m*P
```
